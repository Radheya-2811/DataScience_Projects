```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('compressed_data.csv')
```

```python
df.head()
```

```python
df.columns
```

## checking missing values

```python
print(df.isnull().sum())
```

```python
df.info()
```

```python
df['last review']=pd.to_datetime(df['last review'],errors='coerce')
```

```python
df.info()
```

```python
df.fillna({'reviews per month':0 ,'last review':df['last review'].min()},inplace=True)
```

```python
df.dropna(subset={'NAME','host name'},inplace=True)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.drop(['license','house_rules'],errors='ignore',axis=1 ,inplace=True)
```

```python
df.info()
```

```python
df.head()
```

```python
df['price']=df['price'].replace('[\$,]','',regex=True).astype(float)
df['service fee']=df['service fee'].replace('[\$,]','',regex=True).astype(float)
```

```python
df.head()
```

### removing duplicates

```python
df.drop_duplicates(inplace=True)
```

```python
df.info()
```

### Descriptive Statistics

```python
df.describe()
```

# Visualisation

#### 1. What is the distribution of listing prices



```python
plt.figure(figsize=(10,6))
sns.histplot(df['price'],bins=50,kde=True)
plt.title("Distribution of listing prices")
plt.xlabel('Price $')
plt.ylabel('Frequency')
plt.show()
```

the histogram shows a fairly even distribution of listing prices across different 
price ranges,indicating no particular concentration of listings in any specific range.
The KDE line helps visualize this even spread more clearly.
confirming that the dataset contains listings with wide variety of prices

#### 2. How are different room types distributed

```python
df['room type']
```

```python
plt.figure(figsize=(8,5))
sns.countplot(x='room type',data=df)
plt.title("Room Type Distribution")
plt.xlabel("Room type")
plt.show()
```

#### How are listing distributed across different neighborhoods

```python
df['neighbourhood'].value_counts()
```

```python
plt.figure(figsize=(12,8))
sns.countplot(y='neighbourhood group',data=df,color='violet', order=df['neighbourhood group'].value_counts().index)
plt.xlabel('Count')
plt.show()
```

#### What is the relationship between price and room type

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='room type',y='price',hue='room type',data=df,palette='Set1')
plt.legend(title='Room Type')
plt.show()
```

#### How has the number of reviews changed over time

```python
df['last review']=pd.to_datetime(df['last review'])
reviews_over_time=df.groupby(df['last review'].dt.to_period('M')).size()

plt.figure(figsize=(10,6))
reviews_over_time.plot(kind='line',color='red')
plt.title("Number of reviews over time")
plt.xlabel('Date')
plt.ylabel('Number of reviews')
plt.show()
```
