```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df=pd.read_csv('Banking.csv')
df.head()
```

```python
#Generating Descriptive statistic for the dataframe
df.describe()
```

```python
#Converting Estimated income numerical data to categorical
bins=[0,100000,300000,float('inf')]
labels=['Low','Med','High']

df['Income Band']=pd.cut(df['Estimated Income'],bins=bins,labels=labels,right=False)
```

```python
df['Income Band'].value_counts().plot(kind='bar')
```

```python
#Examine the distribution of unique categories in categorical column
categorical_cols=df[['Nationality','Fee Structure','Loyalty Classification','BRId','GenderId','IAId','Income Band','Properties Owned','Risk Weighting','Amount of Credit Cards']].columns

for cols in categorical_cols:
    print(f"Value count for '{cols}':")
    display(df[cols].value_counts())
```

```python
for i, predictor in enumerate(categorical_cols):
    plt.figure(i)
    sns.countplot(data=df,x=predictor,hue='GenderId')
```
